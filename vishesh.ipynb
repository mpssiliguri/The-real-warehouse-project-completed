```python

import mysql.connector as Mycon
```

```python
a = Mycon.connect(host="localhost", user="root",
                  password="mps@2014", database="warehouse")
```

```python
cursor = a.cursor()
```

```python
cursor.execute("""CREATE TABLE customers (
customer_id INT PRIMARY KEY,
customer_name VARCHAR(50),
phone VARCHAR(15),
address VARCHAR(100)
)""")
```

```python
# INSERT
b = int(input("Enter Customer ID: "))
c = input("Enter Customer Name: ")
d = input("Enter Phone Number: ")
e = input("Enter Address: ")

query = "INSERT INTO customers VALUES (%s, %s, %s, %s)"
value = (b, c, d, e)

cursor.execute(query, value)
a.commit()

print("Customer inserted successfully!")
```

```python
# SELECT
cursor.execute("SELECT * FROM customers")
result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# UPDATE
b = int(input("Enter Customer ID: "))
c = input("Enter New Customer Name: ")
d = input("Enter New Phone Number: ")
e = input("Enter New Address: ")

query = """UPDATE customers
SET customer_name=%s, phone=%s, address=%s
WHERE customer_id=%s"""

value = (c, d, e, b)

cursor.execute(query, value)
a.commit()

print("Customer updated successfully!")
```

```python
# DELETE
b = int(input("Enter Customer ID: "))

query = "DELETE FROM customers WHERE customer_id=%s"
value = (b,)

cursor.execute(query, value)
a.commit()

print("Customer deleted successfully!")

cursor.close()
a.close()
```
