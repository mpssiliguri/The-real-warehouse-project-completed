```python

import mysql.connector as Mycon
```

```python
a = Mycon.connect(host="localhost", user="root",
                  password="mps@2014", database="warehouse")
```

```python
cursor = a.cursor()
```

```python
cursor.execute("""CREATE TABLE suppliers (
supplier_id INT PRIMARY KEY,
supplier_name VARCHAR(50),
phone VARCHAR(15),
address VARCHAR(100)
)""")
```

```python
# INSERT
b = int(input("Enter Supplier ID: "))
c = input("Enter Supplier Name: ")
d = input("Enter Phone Number: ")
e = input("Enter Address: ")

query = "INSERT INTO suppliers VALUES (%s, %s, %s, %s)"
value = (b, c, d, e)

cursor.execute(query, value)
a.commit()

print("Supplier inserted successfully!")
```

```python
# SELECT
cursor.execute("SELECT * FROM suppliers")
result = cursor.fetchall()

for row in result:
    print(row)
```

```python
# UPDATE
b = int(input("Enter Supplier ID: "))
c = input("Enter New Supplier Name: ")
d = input("Enter New Phone Number: ")
e = input("Enter New Address: ")

query = """UPDATE suppliers
SET supplier_name=%s, phone=%s, address=%s
WHERE supplier_id=%s"""

value = (c, d, e, b)

cursor.execute(query, value)
a.commit()

print("Supplier updated successfully!")
```

```python
# DELETE
b = int(input("Enter Supplier ID: "))

query = "DELETE FROM suppliers WHERE supplier_id=%s"
value = (b,)

cursor.execute(query, value)
a.commit()

print("Supplier deleted successfully!")

cursor.close()
a.close()
```
