```python
import mysql.connector as Mycon
```

```python
a=Mycon.connect(host="localhost",user="root",password="mps@2014",database="warehouse")
cursor=a.cursor()
```

```python
cursor.execute("create database warehouse")
```

```python
cursor.execute("""CREATE TABLE products (`s.no` INT PRIMARY KEY,product_name VARCHAR(50),quantity INT)""")
```

```python
b = int(input("Enter Serial Number:"))
c = input("Enter Product Name:")
d = int(input("Enter Quantity:"))
query = "INSERT INTO products (`s.no`,product_name,quantity) VALUES (%s, %s, %s)"
value = (b, c, d)
cursor.execute(query, value)
a.commit()
print("Product inserted successfully!")
```

```python
query = "SELECT * FROM products"
cursor.execute(query)

result = cursor.fetchall()

for row in result:
    print(row)
```
